# Diabetes Prediction using Machine Learning

## Data Loading and Initial Exploration

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn import svm
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
```

```python
diabetes = pd.read_csv('./diabetes.csv')
```

## Data Quality & Preprocessing

```python
zero_not_accepted = ['Glucose', 'BloodPressure', 'SkinThickness', 'Insulin', 'BMI']
```

```python
diabetes[zero_not_accepted] = diabetes[zero_not_accepted].replace(0, np.nan)
```

```python
diabetes.isnull().sum()
```

```python
# Fill NaN with median (more robust than mean for medical data)
diabetes.fillna(diabetes.median(), inplace=True)
```

```python
diabetes.isnull().sum()
```

```python
X = diabetes.drop('Outcome', axis=1)
y = diabetes['Outcome']
```

```python
# Feature Scaling 

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

X = pd.DataFrame(X_scaled, columns=X.columns)
```

```python
print(f"X shape: {X.shape}")
print(f"y shape: {y.shape}")
print(f"\nFirst few rows of scaled X:")
print(X.head())
```

## Train-Test Split

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=2, stratify=y)
```

```python
print(f"X_train shape: {X_train.shape}")
print(f"X_test shape: {X_test.shape}")
print(f"y_train shape: {y_train.shape}")
print(f"y_test shape: {y_test.shape}")
```

```python
print("\nTraining set class distribution:")
print(y_train.value_counts())
print(f"Percentage: {y_train.value_counts(normalize=True) * 100}")

print("\nTest set class distribution:")
print(y_test.value_counts())
print(f"Percentage: {y_test.value_counts(normalize=True) * 100}")
```

## Model Training

```python
classifier = svm.SVC(kernel='linear')
```

```python
classifier.fit(X_train, y_train)
```

```python
print("SVM model training completed successfully!")
print(f"Kernel: {classifier.kernel}")
print(f"Number of support vectors: {classifier.n_support_}")
print(f"Total support vectors: {sum(classifier.n_support_)}")
```

## Model Evaluation

```python
X_train_predictions = classifier.predict(X_train)
X_test_prediction = classifier.predict(X_test)
```

```python
training_accuracy  = accuracy_score(y_train, X_train_predictions)
test_accuracy = accuracy_score(y_test, X_test_prediction)

print(f"Training Accuracy: {training_accuracy :.4f}")
print(f"Test Accuracy: {test_accuracy:.4f}")
```

```python
print("\nConfusion Matrix (Test Set):")
print(confusion_matrix(y_test, X_test_prediction))
```

```python
print("\nClassification Report (Test Set):")
print(classification_report(y_test, X_test_prediction, target_names=['No Diabetes (0)', 'Has Diabetes (1)']))
```

## Making a predictive system 

```python
# Input data (Pregnancies, Glucose, BloodPressure, SkinThickness, Insulin, BMI, DiabetesPedigreeFunction, Age)
input_data = (5, 166, 72, 19, 175, 25.8, 0.587, 51)

# Convert to numpy array and reshape
input_data_as_numpy_array = np.asarray(input_data)
input_data_reshaped = input_data_as_numpy_array.reshape(1, -1)

# Create DataFrame with column names
input_df = pd.DataFrame(input_data_reshaped, columns=X.columns)

# Standardize the input data (returns numpy array)
std_data = scaler.transform(input_df)

# Convert scaled data back to DataFrame with column names - THIS IS THE KEY!
std_data_df = pd.DataFrame(std_data, columns=X.columns)

print("Standardized data:")
print(std_data_df)

# Make prediction on DataFrame (no warnings!)
prediction = classifier.predict(std_data_df)
print(f"\nPrediction: {prediction[0]}")

# Interpret result
if prediction[0] == 0:
    print('The person is not diabetic')
else:
    print('The person is diabetic')
```
